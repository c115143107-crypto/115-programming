使用 pyautogui來完成小恐龍的遊戲機器人。
小恐龍在 chorme 中使用 `chrome://dino` 啟動
邏輯:
- 不斷監視畫面
- 讀取某個位置的顏色
- 若此位置上出現(83,83,83)的顏色
- 跳起來

```python
import pyautogui as pyg
pyg.sleep(2) #sleep 2 sec
pyg.press('space')#啟動

#無窮迴圈
while True:
    im= pyg.screenshot()
    obs=False #假設沒有仙人掌 
    for i in range(100):
        c=im.getpixel((245+i,580+i))
        if c ==(83,83,83):
            obs=True
            break
    if obs==True:
        pyg.press('space')
```

```python
import pyautogui as pyg

# 先取得當前螢幕截圖來檢查特定位置的顏色
im_check = pyg.screenshot()
pos_color = im_check.getpixel((1118, 446))

while True:
    im = pyg.screenshot()
    obs = False  # 假設沒有仙人掌

    # 根據 (1118, 446) 的顏色來決定判斷條件
    if pos_color == (255, 255, 255):
        for i in range(100):
            c = im.getpixel((245 + i, 580 + i))
            if c == (83, 83, 83):
                obs = True
                break
    elif pos_color == (0, 0, 0):
        for i in range(100):
            c = im.getpixel((245 + i, 580 + i))
            if c == (0, 0, 0):
                obs = True
                break

    if obs == True:
        pyg.press('space')
```

```python
import pyautogui as pyg

# 取得當前畫面的截圖來檢查背景顏色
im_check = pyg.screenshot()
bg_color = im_check.getpixel((1118, 446))

while True:
    im = pyg.screenshot()
    obs = False  # 假設沒有障礙物

    # 情況一：當螢幕截圖時 position(1118, 446) 的顏色為 (255, 255, 255)（白色背景）
    if bg_color == (255, 255, 255):
        for i in range(100):
            c = im.getpixel((245 + i, 580 + i))
            if c == (83, 83, 83):
                obs = True
                break

    # 情況二：當螢幕截圖時 position(1118, 446) 的顏色為 (0, 0, 0)（黑色背景）
    elif bg_color == (0, 0, 0):
        for i in range(100):
            c = im.getpixel((167 + i, 580 + i))
            if c == (172, 172, 172):
                obs = True
                break

    # 只要其中一個條件成立並偵測到障礙物，就會按下空白鍵
    if obs:
        pyg.press('space')
```

```python
import pyautogui as pyg

while True:
    im = pyg.screenshot()
    
    # 條件一：檢查特定區域是否有顏色 (83, 83, 83) 的障礙物
    obs1 = False
    for i in range(100):
        c1 = im.getpixel((245 + i, 580 + i))
        if c1 == (83, 83, 83):
            obs1 = True
            break
            
    # 條件二：檢查座標 (1118, 446) 的顏色是否從 (255, 255, 255) 轉變為 (0, 0, 0)
    # 透過讀取當前顏色並判斷是否為 (0, 0, 0) 來實現轉變觸發
    c2 = im.getpixel((1118, 446))
    obs2 = (c2 == (0, 0, 0))
    
    # 兩個條件只要有一個成立（or），就按下空白鍵
    if obs1 or obs2:
        pyg.press('space')
```

```python
import pyautogui as pyg
# 記得需要引入 time 模組來控制迴圈速度（避免 CPU 佔用過高）
import time

while True:
    im = pyg.screenshot()
    obs = False  # 假設沒有障礙物
    
    # 檢查範圍內的像素
    for i in range(100):
        c = im.getpixel((245 + i, 580 + i))
        
        # 只要偵測到原本的顏色 (83,83,83) 或新增的顏色 (0,0,0) 其中之一
        if c == (83, 83, 83) or c == (0, 0, 0):
            obs = True
            break
            
    # 如果任一條件成立，就按下空白鍵
    if obs == True:
        pyg.press('space')
        
    # 建議加入微小的停頓，避免迴圈跑太快導致 CPU 資源耗盡
    time.sleep(0.01)
```

```python
import pyautogui as pyg
import time

while True:
    im = pyg.screenshot()
    obs = False  # 假設沒有障礙物
    
    # 檢查 1：尋找顏色為 (83, 83, 83) 的像素
    for i in range(100):
        c = im.getpixel((245 + i, 580 + i))
        if c == (83, 83, 83):
            obs = True
            break
            
    # 檢查 2：如果第一個條件還沒成立，再檢查顏色為 (0, 0, 0) 的像素
    if not obs:
        for i in range(100):
            c = im.getpixel((245 + i, 580 + i))
            if c == (0, 0, 0):
                obs = True
                break
                
    # 只要其中一個條件成立，就按下空白鍵
    if obs:
        pyg.press('space')
        
    # 稍微延遲避免佔用過多 CPU 資源（可依需求調整或拿掉）
    time.sleep(0.01)
```
