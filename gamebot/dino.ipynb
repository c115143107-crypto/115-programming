使用 pyautogui來完成小恐龍的遊戲機器人。
小恐龍在 chorme 中使用 `chrome://dino` 啟動
邏輯:
- 不斷監視畫面
- 讀取某個位置的顏色
- 若此位置上出現(83,83,83)的顏色
- 跳起來

```python
import pyautogui as pyg
pyg.sleep(2) #sleep 2 sec
pyg.press('space')#啟動

#無窮迴圈
while True:
    im= pyg.screenshot()
    obs=False #假設沒有仙人掌 
    for i in range(100):
        c=im.getpixel((214+i,588+i))
        if c ==(83,83,83):
            obs=True
            break
    if obs==True:
        pyg.press('space')
```

```python
import time
import pyautogui as pyg

pyg.sleep(2)  # 等待 2 秒讓你切換到遊戲畫面
pyg.press('space')  # 啟動遊戲

# 無窮迴圈
while True:
  im = pyg.screenshot()
  obs = False  # 假設沒有障礙物

  # 偵測範圍設定
  start_x = 214
  width = 70
  
  # 修正 y_pos 指定方式 (如果需要偵測兩個不同高度，可以分開掃描，這裡先指定一個基準高度或用變數儲存)
  # 建議先以你原本的 588 或其他高度為準
  y_pos = 660  # 依你的畫面高度調整

  for i in range(width):
    # 避免 y_pos 寫錯，確保抓取的是正確像素
    c = im.getpixel((start_x + i, y_pos))

    # 檢查是否符合指定的兩種顏色之一
    if c == (83, 83, 83) or c == (172, 172, 172):
      obs = True
      break

  if obs == True:
    # 改用 mouseDown 與 mouseUp（或對應的 key 動作，pyg 也能用 keydown/keyup）
    # 讓空白鍵「按住」約 0.15~0.2 秒，小恐龍就會跳到最高！
    pyg.keyDown('space')
    time.sleep(0.18)  # 控制跳躍高度的時間 (可依需求微調 0.15 ~ 0.25)
    pyg.keyUp('space')
    
    # 避免在同一個障礙物上重複連續觸發
    time.sleep(0.1)
```

```python
import time
import pyautogui as pyg

pyg.sleep(2)  # 等待 2 秒讓你切換到遊戲畫面
pyg.press('space')  # 啟動遊戲

# 無窮迴圈
while True:
  im = pyg.screenshot()
  obs = False  # 假設沒有障礙物

  # 根據你的需求設定範圍：
  # X 範圍：214 到 284 (總寬度 70)
  # Y 範圍：587 到 632
  start_x = 220
  end_x = 284
  start_y = 587
  end_y = 680

  # 使用雙重迴圈掃描指定的 X 與 Y 區間
  for x in range(start_x, end_x + 1):
    for y in range(start_y, end_y + 1):
      c = im.getpixel((x, y))

      # 檢查是否符合指定的兩種顏色之一
      if c == (83, 83, 83) or c == (172, 172, 172):
        obs = True
        break  # 跳出內層 y 迴圈

    if obs:
      break  # 跳出外層 x 迴圈

  # 如果在指定範圍內偵測到目標顏色，執行跳躍
  if obs:
    pyg.keyDown('space')
    time.sleep(0.18)  # 控制跳躍高度的時間
    pyg.keyUp('space')

    # 避免在同一個障礙物上連續觸發，稍微緩衝一下
    time.sleep(0.1)
```

```python
import time
import pyautogui as pyg

pyg.sleep(2)  # 等待 2 秒讓你切換到遊戲畫面
pyg.press('space')  # 啟動遊戲

# 無窮迴圈
while True:
  im = pyg.screenshot()
  obs = False
  jump_type = 'normal'  # 預設為正常跳

  # 定義掃描範圍（將 end_y 擴大，涵蓋到 652 以下的區域）
  start_x = 224
  end_x = 300
  start_y = 583
  end_y = 655  # 💡 擴大 y 的掃描範圍，讓下方大於 652 的區域也能被偵測到

  # 掃描區域內的像素
  for x in range(start_x, end_x + 1):
    for y in range(start_y, end_y + 1):
      c = im.getpixel((x, y))

      # 檢查指定顏色
      if c == (83, 83, 83) or c == (172, 172, 172):
        obs = True

        # 根據 y 座標決定跳躍方式：
        # 如果 y 小於或等於 652（位置較高/較遠） -> 跳到最高
        # 如果 y 大於 652（位置較低/較近） -> 正常跳
        if y <= 652:
          jump_type = 'high'
        else:
          jump_type = 'normal'

        break
    if obs:
      break

  if obs:
    if jump_type == 'high':
      # 滿足 y <= 652：跳到最高（按住空白鍵 0.18 秒）
      pyg.keyDown('space')
      time.sleep(0.18)
      pyg.keyUp('space')
    else:
      # 滿足 y > 652：正常跳
      pyg.press('space')

    # 避免重複連續觸發
    time.sleep(0.15)
```

```python
import time
import pyautogui as pyg

pyg.sleep(2)  # 等待 2 秒讓你切換到遊戲畫面
pyg.press('space')  # 啟動遊戲

# 無窮迴圈
while True:
  im = pyg.screenshot()
  obs = False
  jump_type = 'normal'  # 預設為正常跳

  # 定義掃描範圍
  start_x = 253
  end_x = 310
  start_y = 587
  end_y = 655  # 涵蓋較大範圍

  # 掃描區域內的像素
  for x in range(start_x, end_x + 1):
    for y in range(start_y, end_y + 1):
      r, g, b = im.getpixel((x, y))

      # 檢查是否為介於 83 到 172 之間的灰色/障礙物色彩區間
      # (Chrome 遊戲中障礙物通常是 R=G=B 的灰階色，所以可以檢查各分量是否都在 83~172 之間)
      if (83 <= r <= 172) and (83 <= g <= 172) and (83 <= b <= 172):
        obs = True

        # 根據 y 座標決定跳躍方式
        if y <= 652:
          jump_type = 'high'
        else:
          jump_type = 'normal'

        break
    if obs:
      break

  if obs:
    if jump_type == 'high':
      # 滿足 y <= 652：跳到最高（按住空白鍵 0.18 秒）
      pyg.keyDown('space')
      time.sleep(0.18)
      pyg.keyUp('space')
    else:
      # 滿足 y > 652：正常跳
      pyg.press('space')

    # 避免重複連續觸發
    time.sleep(0.15)
```
