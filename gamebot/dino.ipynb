使用 pyautogui來完成小恐龍的遊戲機器人。
小恐龍在 chorme 中使用 `chrome://dino` 啟動
邏輯:
- 不斷監視畫面
- 讀取某個位置的顏色
- 若此位置上出現(83,83,83)的顏色
- 跳起來

```python
import pyautogui as pyg
pyg.sleep(2) #sleep 2 sec
pyg.press('space')#啟動

#無窮迴圈
while True:
    im= pyg.screenshot()
    obs=False #假設沒有仙人掌 
    for i in range(100):
        c=im.getpixel((245+i,580+i))
        if c ==(83,83,83):
            obs=True
            break
    if obs==True:
        pyg.press('space')
```

```python
import pyautogui as pyg
import time

# 確保 pyautogui 有預設的防呆停頓時間
pyg.FAILSAFE = True

# 用來記錄上一幀畫面的平均亮度，以便偵測日夜瞬間切換
last_brightness = None

while True:
    im = pyg.screenshot()
    obs = False  # 假設沒有障礙物
    
    # 1. 原有的障礙物顏色檢查 (83,83,83 或 172,172,172 或 0,0,0)
    for i in range(100):
        c = im.getpixel((245 + i, 580 + i))
        if c == (83, 83, 83) or c == (172, 172, 172) or c == (0, 0, 0):
            obs = True
            break  
        pyg.press('space')                                                                                                                                                                                                                                                                                                                                                                     
            
    # 2. 新增：日夜瞬間切換（畫面整體明暗劇烈變化）偵測
    # 為了效能，我們縮小圖片來計算整體平均亮度
    small_im = im.resize((50, 50))
    pixels = list(small_im.getdata())
    # 計算平均亮度 (R+G+B 的平均值)
    current_brightness = sum(sum(p) / 3 for p in pixels) / len(pixels)
    
    if last_brightness is not None:
        # 計算這一幀跟上一幀的亮度差，如果變化量太大（代表日夜瞬間切換）
        brightness_diff = abs(current_brightness - last_brightness)
        
        # 門檻值設定為 30（可依實際畫面反轉程度微調）
        if brightness_diff > 30:
            obs = True
            
    # 更新上一幀的亮度記錄
    last_brightness = current_brightness

    # 如果任一條件成立（偵測到障礙物 或 發生日夜切換閃爍），立刻按下空白鍵
    if obs == True:
        pyg.press('space')
        
    # 稍微暫停，確保效能流暢
    time.sleep(0.01)
```
